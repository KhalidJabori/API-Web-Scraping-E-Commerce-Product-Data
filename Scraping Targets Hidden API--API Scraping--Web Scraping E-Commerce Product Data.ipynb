```python
products_title=[]
price=[]
brand=[]
rating_avg=[]
rating_count=[]

for result in results_items:
    try:      
        products_title.append(result['item']['product_description']['title'].replace('&#8482;',''))
    except:   
        products_title.append('')
        
    try:    
        price.append(result['price']['current_retail'])
    except:   
        price.append('')
      
    try:
        brand.append(result['item']['primary_brand']['name'])
    except:   
        brand.append('')
        
    try:   
        rating_avg.append(result['ratings_and_reviews']['statistics']['rating']['average'])
    except:   
        rating_avg.append('')  
        
        
    try:
        rating_count.append(result['ratings_and_reviews']['statistics']['rating']['count'])
    except:   
        rating_count.append('')
```

```python
target_df = pd.DataFrame({'Products_title':products_title,'Price':price,'brand':brand,'rating_avg':rating_avg,'rating_count':rating_count})
```

### Multiple Pages

```python
url = "https://redsky.target.com/redsky_aggregations/v1/web/plp_search_v2"
products_title=[]
price=[]
brand=[]
rating_avg=[]
rating_count=[]

for i in range(0,240,24):

    params = {
        "key": "9f36aeafbe60771e321a7cc95a78140772ab3e96",
        "category": "5xt7a",
        "channel": "WEB",
        "count": "24",
        "default_purchasability_filter": "true",
        "include_dmc_dmr": "true",
        "include_review_summarization": "false",
        "new_search": "false",
        "offset": str(i),
        "page": "/c/5xt7a",
        "platform": "desktop",
        "pricing_store_id": "1771",
        "scheduled_delivery_store_id": "1771",
        "spellcheck": "true",
        "store_ids": "1771,1768,1113,3374,1792",
        "useragent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36",
        "visitor_id": "019653D8D8C80201910925902E2CD857",
        "zip": "52404"
    }

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36",
        "Referer": "https://www.target.com/c/weights-strength-training-exercise-fitness-sports-outdoors/-/N-5xt7a?moveTo=product-list-grid"
    }
    
    #response
    response = requests.get(url, params=params, headers=headers)
    
    results_json = response.json()
    
    results_items=results_json['data']['search']['products']

    
    for result in results_items:
        try:      
            products_title.append(result['item']['product_description']['title'].replace('&#8482;',''))
        except:   
            products_title.append('')

        try:    
            price.append(result['price']['current_retail'])
        except:   
            price.append('')

        try:
            brand.append(result['item']['primary_brand']['name'])
        except:   
            brand.append('')

        try:   
            rating_avg.append(result['ratings_and_reviews']['statistics']['rating']['average'])
        except:   
            rating_avg.append('')  


        try:
            rating_count.append(result['ratings_and_reviews']['statistics']['rating']['count'])
        except:   
            rating_count.append('')
```

```python
target_df = pd.DataFrame({'Products_title':products_title,'Price':price,'brand':brand,'rating_avg':rating_avg,'rating_count':rating_count})
```
